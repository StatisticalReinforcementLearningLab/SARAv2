```python
from cairosvg import svg2png



for hour in range(1,13):
    for minute in [0, 15, 30, 45]:
    
        print(f"hour:minute == {hour}:{minute}")
        hoursAngle = 360 * hour / 12 + minute / 2;
        minuteAngle = 360 * minute / 60;

        svg_code = f"""
            <svg width="100" height="100">
                <g>
                    <circle id="circle" style="stroke: #FFF; stroke-width: 0px; fill:#27ae60" cx="50" cy="50" r="50"></circle>
                </g>
                <g>
                    <line x1="50" y1="50" x2="50" y2="20" transform="rotate({hoursAngle} 50 50)" style="stroke-width: 10px; stroke: #fffbf9;" id="hourhand">
                    </line>
                    <line x1="50" y1="50" x2="50" y2="5" transform="rotate({minuteAngle} 50 50)" style="stroke-width: 5px; stroke: #fdfdfd;" id="minutehand">
                    </line>
                </g>
                <circle id="center" style="fill:#128A86; stroke: #C1EFED; stroke-width: 2px;" cx="50" cy="50" r="7"></circle>
            </svg>
        """

        svg2png(bytestring=svg_code, write_to=f'output-{hour}-{minute}.png')
```
